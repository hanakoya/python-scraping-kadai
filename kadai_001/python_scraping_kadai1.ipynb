```python
!pip install requests beautifulsoup4
import requests
from bs4 import BeautifulSoup
url='https://news.yahoo.co.jp/articles/64a72101d9b7bd8a07676c7321294841ea0bdaa6'
response=requests.get(url)
soup=BeautifulSoup(response.text, 'html.parser')
content_element=soup.select_one('#uamods article, #uamods .article_body')
if content_element:
    # キャプション等を除外するため、記事要素内のpタグ（本文段落）テキストのみを抽出して結合します
    paragraphs = content_element.find_all('p')
    content_text = '\n'.join([p.text for p in paragraphs if p.text.strip()])
else:
    content_text = "本文が見つかりませんでした。"
print(content_text)
```
